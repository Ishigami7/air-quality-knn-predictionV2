## 📊 Data Loading and Initial Exploration

```python
# Option 1: Load your own data
# Uncomment and modify the path below to load your own data
# data_processor = DataProcessor()
# data = data_processor.load_data('path/to/your/air_quality_data.csv')

# Option 2: Use sample data for demonstration
print("🔄 Creating sample air quality data for demonstration...")
data = create_sample_data(n_samples=1000)

print(f"📊 Data shape: {data.shape}")
print(f"📋 Columns: {list(data.columns)}")
print("\n📈 First 5 rows:")
data.head()
```

```python
# Initialize data processor and analyzer
data_processor = DataProcessor()
data_processor.data = data

# Data validation
print("🔍 COMPREHENSIVE DATA VALIDATION")
print("=" * 50)

validation_results = data_processor.validate_data()

print(f"📊 Data Types:")
for col, dtype in validation_results['data_types'].items():
    print(f"   {col}: {dtype}")

print(f"\n🔍 Missing Values:")
if validation_results['missing_values']:
    for col, count in validation_results['missing_values'].items():
        percentage = (count / len(data)) * 100
        print(f"   {col}: {count} ({percentage:.2f}%)")
else:
    print("   No missing values found!")

print(f"\n🔄 Duplicate Rows: {validation_results['duplicate_rows']}")

if validation_results['range_issues']:
    print(f"\n⚠️ Range Issues:")
    for col, issue in validation_results['range_issues'].items():
        print(f"   {col}: {issue}")
else:
    print("\n✅ No range issues detected!")
```

## 🧹 Data Preprocessing

```python
print("🧹 DATA PREPROCESSING")
print("=" * 50)

# Handle missing values
data_cleaned = data_processor.handle_missing_values(strategy='mean')
print(f"✅ Missing values handled using mean strategy")

# Create air quality categories
data_with_categories = data_processor.create_air_quality_categories()
data_processor.data = data_with_categories
print(f"✅ Air quality categories created")

# Feature engineering
data_engineered = data_processor.feature_engineering(include_interactions=True)
data_processor.data = data_engineered
print(f"✅ Feature engineering completed")

# Display category distribution
print(f"\n🎯 Air Quality Category Distribution:")
category_counts = data_engineered['AirQuality_Category'].value_counts()
for category, count in category_counts.items():
    percentage = (count / len(data_engineered)) * 100
    print(f"   {category}: {count} ({percentage:.1f}%)")

print(f"\n📊 Final dataset shape: {data_engineered.shape}")
print(f"📋 New columns added: {[col for col in data_engineered.columns if col not in data.columns]}")
```

## 📈 Exploratory Data Analysis (EDA)

```python
print("📈 EXPLORATORY DATA ANALYSIS")
print("=" * 50)

# Initialize analyzer
analyzer = AirQualityAnalyzer(data_engineered)

# Descriptive statistics
stats = analyzer.descriptive_statistics()

print("📊 Descriptive Statistics:")
display(stats['basic'])

print("\n📊 Skewness and Kurtosis:")
for feature in analyzer.feature_columns:
    if feature in data_engineered.columns:
        skew_val = stats['skewness'].get(feature, 'N/A')
        kurt_val = stats['kurtosis'].get(feature, 'N/A')
        print(f"   {feature}: Skewness = {skew_val:.3f}, Kurtosis = {kurt_val:.3f}")
```

```python
# Outlier detection
print("🔍 OUTLIER DETECTION")
print("=" * 30)

outliers_info = data_processor.detect_outliers(method='iqr', threshold=1.5)

for feature, info in outliers_info.items():
    print(f"📊 {feature}:")
    print(f"   Outliers: {info['count']} ({info['percentage']:.2f}%)")
    if 'bounds' in info:
        print(f"   Normal range: {info['bounds'][0]:.2f} to {info['bounds'][1]:.2f}")
    print()
```

```python
# Correlation analysis
print("🔗 CORRELATION ANALYSIS")
print("=" * 30)

correlation_results = analyzer.correlation_analysis()

if 'high_correlation_pairs' in correlation_results:
    print("🔗 Highly Correlated Feature Pairs (|correlation| > 0.7):")
    for pair in correlation_results['high_correlation_pairs']:
        print(f"   {pair['feature1']} ↔ {pair['feature2']}: {pair['correlation']:.3f}")
else:
    print("✅ No highly correlated feature pairs found.")
```

## 🤖 Model Development

```python
print("🤖 KNN MODEL DEVELOPMENT")
print("=" * 50)

# Prepare features and target
feature_columns = ['PM10', 'PM25', 'SO2', 'CO', 'O3', 'NO2']
available_features = [col for col in feature_columns if col in data_engineered.columns]

# Add engineered features if available
engineered_features = ['PM25_PM10_Ratio', 'Combined_Pollution_Index']
for feat in engineered_features:
    if feat in data_engineered.columns:
        available_features.append(feat)

X = data_engineered[available_features]
y = data_engineered['AirQuality_Category']

print(f"📊 Features used: {available_features}")
print(f"🎯 Target variable: AirQuality_Category")
print(f"📈 Dataset size: {len(X)} samples, {len(available_features)} features")

# Initialize model
model = KNNAirQualityModel(task_type='classification')
print(f"✅ KNN model initialized for classification")
```

## 🔧 Hyperparameter Tuning

```python
print("🔧 HYPERPARAMETER TUNING")
print("=" * 50)

# Perform hyperparameter tuning
tuning_results = model.hyperparameter_tuning(X, y, cv_folds=5, scoring='accuracy')

print(f"\n🎯 Hyperparameter Tuning Results:")
print(f"   Best Parameters: {tuning_results['best_params']}")
print(f"   Best CV Score: {tuning_results['best_score']:.4f}")
print(f"   Total Combinations Tested: {tuning_results['n_combinations_tested']}")
```

```python
# Visualize hyperparameter tuning results
if 'cv_results' in tuning_results:
    fig_tuning = visualizer.plot_hyperparameter_tuning(tuning_results['cv_results'], 'n_neighbors')
    if fig_tuning:
        plt.show()
```

## 🚀 Model Training and Evaluation

```python
print("🚀 MODEL TRAINING")
print("=" * 50)

# Train the model
training_results = model.train(X, y, test_size=0.2, random_state=42)

print(f"\n📊 Training Results:")
print(f"   Training Set Size: {training_results['train_size']}")
print(f"   Test Set Size: {training_results['test_size']}")
print(f"   Training Accuracy: {training_results['train_accuracy']:.4f}")
print(f"   Test Accuracy: {training_results['test_accuracy']:.4f}")

print(f"\n📈 Classification Report:")
classification_report = training_results['classification_report']
for class_name, metrics in classification_report.items():
    if isinstance(metrics, dict) and class_name not in ['accuracy', 'macro avg', 'weighted avg']:
        print(f"   {class_name}:")
        print(f"     Precision: {metrics['precision']:.3f}")
        print(f"     Recall: {metrics['recall']:.3f}")
        print(f"     F1-Score: {metrics['f1-score']:.3f}")

print(f"\n📊 Overall Metrics:")
macro_avg = classification_report['macro avg']
print(f"   Macro Avg Precision: {macro_avg['precision']:.3f}")
print(f"   Macro Avg Recall: {macro_avg['recall']:.3f}")
print(f"   Macro Avg F1-Score: {macro_avg['f1-score']:.3f}")
```

```python
# Cross-validation
print("🔄 CROSS-VALIDATION")
print("=" * 30)

cv_results = model.cross_validate(X, y, cv_folds=5)

for metric, results in cv_results.items():
    print(f"📊 {metric.replace('_', ' ').title()}:")
    print(f"   Mean: {results['mean']:.4f} (±{results['std']:.4f})")
    print(f"   Range: [{results['min']:.4f}, {results['max']:.4f}]")
    print()
```

```python
# Learning curves
print("📈 GENERATING LEARNING CURVES")
print("=" * 30)

learning_curve_data = model.learning_curves(X, y)
fig_learning = visualizer.plot_learning_curves(learning_curve_data)
if fig_learning:
    plt.show()
```

```python
# Model performance visualization
fig_performance = visualizer.plot_model_performance(training_results, 'classification')
if fig_performance:
    plt.show()
```

## 🎯 Feature Importance Analysis

```python
print("🎯 FEATURE IMPORTANCE ANALYSIS")
print("=" * 50)

# Calculate feature importance
importance_data = model.feature_importance(X, y, method='permutation')

print(f"📊 Feature Importance (Permutation Method):")
feature_names = importance_data['feature_names']
importance_scores = importance_data['importance_mean']
ranking = importance_data['ranking']

for i, idx in enumerate(ranking[:len(feature_names)]):
    print(f"   {i+1}. {feature_names[idx]}: {importance_scores[idx]:.4f}")

# Plot feature importance
fig_importance = visualizer.plot_feature_importance(importance_data, top_k=min(10, len(feature_names)))
if fig_importance:
    plt.show()
```

## 🔮 Prediction System

```python
print("🔮 PREDICTION SYSTEM DEMONSTRATION")
print("=" * 50)

# Sample predictions
sample_data = X.sample(n=10, random_state=42)
true_labels = y.loc[sample_data.index]

# Make predictions
predictions = model.predict(sample_data)
predictions_with_confidence, confidence_scores = model.predict_with_confidence(sample_data)

print(f"\n📊 Sample Predictions:")
print(f"{'Index':<8} {'True':<12} {'Predicted':<12} {'Confidence':<12} {'Correct':<8}")
print("-" * 60)

for i, (idx, true_val, pred_val, conf) in enumerate(zip(sample_data.index, true_labels, predictions, confidence_scores)):
    correct = "✅" if true_val == pred_val else "❌"
    print(f"{idx:<8} {true_val:<12} {pred_val:<12} {conf:<12.3f} {correct:<8}")

# Calculate sample accuracy
sample_accuracy = (true_labels == predictions).mean()
print(f"\n📈 Sample Accuracy: {sample_accuracy:.3f}")
print(f"📊 Average Confidence: {confidence_scores.mean():.3f}")
```

```python
# Interactive prediction function
def predict_air_quality(pm10, pm25, so2, co, o3, no2):
    """
    Interactive function to predict air quality
    """
    # Create input DataFrame
    input_data = pd.DataFrame({
        'PM10': [pm10],
        'PM25': [pm25],
        'SO2': [so2],
        'CO': [co],
        'O3': [o3],
        'NO2': [no2]
    })
    
    # Add engineered features if model expects them
    if 'PM25_PM10_Ratio' in available_features:
        input_data['PM25_PM10_Ratio'] = input_data['PM25'] / (input_data['PM10'] + 1e-8)
    
    if 'Combined_Pollution_Index' in available_features:
        # Simple normalization for demonstration
        normalized_values = [
            pm10/100, pm25/50, so2/50, co/10, o3/100, no2/50
        ]
        input_data['Combined_Pollution_Index'] = [np.mean(normalized_values)]
    
    # Select only available features
    input_data = input_data[available_features]
    
    # Make prediction
    prediction, confidence = model.predict_with_confidence(input_data)
    
    return prediction[0], confidence[0]

# Example prediction
print("\n🔮 INTERACTIVE PREDICTION EXAMPLE")
print("=" * 40)

# Example values
example_pm10 = 75
example_pm25 = 35
example_so2 = 25
example_co = 2.5
example_o3 = 80
example_no2 = 40

pred_category, pred_confidence = predict_air_quality(
    example_pm10, example_pm25, example_so2, 
    example_co, example_o3, example_no2
)

print(f"📊 Input Values:")
print(f"   PM10: {example_pm10} μg/m³")
print(f"   PM2.5: {example_pm25} μg/m³")
print(f"   SO2: {example_so2} μg/m³")
print(f"   CO: {example_co} mg/m³")
print(f"   O3: {example_o3} μg/m³")
print(f"   NO2: {example_no2} μg/m³")

print(f"\n🎯 Prediction Results:")
print(f"   Air Quality Category: {pred_category}")
print(f"   Confidence Score: {pred_confidence:.3f}")

# Add interpretation
if pred_confidence > 0.8:
    confidence_level = "High"
elif pred_confidence > 0.6:
    confidence_level = "Medium"
else:
    confidence_level = "Low"

print(f"   Confidence Level: {confidence_level}")
```

```python
print("📊 COMPREHENSIVE PERFORMANCE SUMMARY")
print("=" * 60)

# Model summary
model_summary = model.model_summary()

print(f"🤖 Model Information:")
print(f"   Task Type: {model_summary['task_type']}")
print(f"   Model Status: {'Fitted' if model_summary['is_fitted'] else 'Not Fitted'}")
print(f"   Feature Count: {model_summary['feature_count']}")
print(f"   Training Samples: {model_summary.get('training_samples', 'N/A')}")
print(f"   Test Samples: {model_summary.get('test_samples', 'N/A')}")

print(f"\n🎯 Best Hyperparameters:")
for param, value in model_summary['best_hyperparameters'].items():
    print(f"   {param}: {value}")

print(f"\n📈 Performance Metrics:")
print(f"   Training Accuracy: {training_results['train_accuracy']:.4f}")
print(f"   Test Accuracy: {training_results['test_accuracy']:.4f}")
print(f"   Generalization Gap: {abs(training_results['train_accuracy'] - training_results['test_accuracy']):.4f}")

print(f"\n🔗 Feature Ranking (Top 5):")
top_5_features = ranking[:5]
for i, idx in enumerate(top_5_features):
    print(f"   {i+1}. {feature_names[idx]}: {importance_scores[idx]:.4f}")

print(f"\n📊 Cross-Validation Summary:")
accuracy_cv = cv_results.get('accuracy', {})
if accuracy_cv:
    print(f"   Mean CV Accuracy: {accuracy_cv['mean']:.4f} (±{accuracy_cv['std']:.4f})")
    print(f"   CV Score Range: [{accuracy_cv['min']:.4f}, {accuracy_cv['max']:.4f}]")

print(f"\n🎯 Model Reliability Assessment:")
# Calculate reliability metrics
cv_stability = 1 - (accuracy_cv.get('std', 0) / accuracy_cv.get('mean', 1))
generalization_score = 1 - abs(training_results['train_accuracy'] - training_results['test_accuracy'])
overall_reliability = (cv_stability + generalization_score) / 2

print(f"   CV Stability: {cv_stability:.3f}")
print(f"   Generalization Score: {generalization_score:.3f}")
print(f"   Overall Reliability: {overall_reliability:.3f}")

if overall_reliability > 0.9:
    reliability_rating = "Excellent ⭐⭐⭐⭐⭐"
elif overall_reliability > 0.8:
    reliability_rating = "Good ⭐⭐⭐⭐"
elif overall_reliability > 0.7:
    reliability_rating = "Fair ⭐⭐⭐"
else:
    reliability_rating = "Needs Improvement ⭐⭐"

print(f"   Rating: {reliability_rating}")
```
